### Exercise 1.4 *** — StandardScaler vs MinMaxScaler

Try scaling with `MinMaxScaler` as well (scales everything to 0-1). Compare:
- How does MinMaxScaler handle outliers differently than StandardScaler?
- When might you prefer one over the other?

*Hint: StandardScaler is sensitive to outliers because it uses mean/std. MinMaxScaler is even more sensitive because a single extreme value stretches the range.*

````python
# Your code here
# 1.4: StandardScaler vs MinMaxScaler

print("=" * 70)
print("COMPARISON: StandardScaler vs MinMaxScaler")
print("=" * 70)

# We already have X_scaled from StandardScaler above
# Now apply MinMaxScaler
minmax_scaler = MinMaxScaler()
X_minmax = pd.DataFrame(minmax_scaler.fit_transform(X), columns=X.columns)

print("\nStandardScaler statistics (mean=0, std=1):")
print(X_scaled.describe())

print("\nMinMaxScaler statistics (values in 0-1 range):")
print(X_minmax.describe())

# Visual comparison
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Original data
X.boxplot(ax=axes[0, 0])
axes[0, 0].set_title('Original Data (UNSCALED)')
axes[0, 0].set_ylabel('Value')

# StandardScaler
X_scaled.boxplot(ax=axes[0, 1])
axes[0, 1].set_title('StandardScaler (mean=0, std=1)')
axes[0, 1].set_ylabel('Scaled Value')

# MinMaxScaler
X_minmax.boxplot(ax=axes[1, 0])
axes[1, 0].set_title('MinMaxScaler (range 0-1)')
axes[1, 0].set_ylabel('Scaled Value')

# Histogram comparison for one feature (body_mass_g)
feature = 'body_mass_g'
axes[1, 1].hist([X[feature], X_scaled[feature], X_minmax[feature]], 
                label=['Original', 'StandardScaler', 'MinMaxScaler'],
                alpha=0.6, bins=20)
axes[1, 1].set_title(f'Distribution of {feature}')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].legend()

plt.tight_layout()
plt.show()

print("\n" + "=" * 70)
print("KEY DIFFERENCES:")
print("=" * 70)
print("""
StandardScaler (mean=0, std=1):
  ✓ Centers data around zero
  ✓ Preserves the shape of the distribution
  ✓ Sensitive to outliers (uses mean/std which are affected by extremes)
  ✓ Good for: algorithms that assume normally distributed data (e.g., K-Means)

MinMaxScaler (range 0-1):
  ✓ Scales all features to exactly [0, 1]
  ✓ VERY sensitive to outliers (a single extreme value stretches the entire range)
  ✓ All features now have the same min/max bounds
  ✓ Good for: neural networks, image data, when you need bounded values

WHEN TO USE:
- StandardScaler: Default choice for clustering (K-Means, hierarchical)
                 Works well with most algorithms
                 
- MinMaxScaler: When you have outliers you want to "squash"
                When you need values in a specific range [0,1]
                For tree-based models (less sensitive to scale anyway)
""")

# Demonstrate outlier sensitivity
print("\nDemonstrating outlier sensitivity:")
print("Original body_mass_g min/max:", X['body_mass_g'].min(), X['body_mass_g'].max())
print("MinMaxScaler result (0-1):", X_minmax['body_mass_g'].min(), X_minmax['body_mass_g'].max())
print("StandardScaler result:", X_scaled['body_mass_g'].min(), X_scaled['body_mass_g'].max())
print("\n→ MinMaxScaler compresses ALL values to [0,1] range")
print("→ If you had an extreme outlier, it would compress everything else more")
````

#### Exercise 2A.2 ** — The Elbow Method

How do you choose the right number of clusters? The **elbow method** plots the within-cluster sum of squares (inertia) for different values of K:

1. Run K-Means for k = 2, 3, 4, ..., 10
2. For each k, record `kmeans.inertia_`
3. Plot k vs. inertia — look for the "elbow" where adding more clusters stops helping much

What value of k does the elbow suggest?

````python
# Exercise 2A.2: The Elbow Method

print("=" * 70)
print("ELBOW METHOD: Finding optimal k")
print("=" * 70)

# 1. Run K-Means for k = 2 to 10 and record inertia
inertias = []
k_values = range(2, 11)

print("\nFitting K-Means for different values of k...")
for k in k_values:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_temp.fit(X_scaled)
    inertias.append(kmeans_temp.inertia_)
    print(f"  k={k:2d}: inertia = {kmeans_temp.inertia_:8.2f}")

# 2. Plot k vs. inertia to visualize the elbow
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(k_values, inertias, 'bo-', linewidth=2, markersize=8, label='Inertia')
ax.set_xlabel('Number of Clusters (k)', fontsize=12)
ax.set_ylabel('Inertia (Within-cluster sum of squares)', fontsize=12)
ax.set_title('Elbow Method - Finding Optimal k', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)

# Highlight our initial choice (k=3)
ax.plot(3, inertias[1], 'r*', markersize=20, label='Our choice (k=3)', zorder=5)

plt.tight_layout()
plt.show()

# 3. Analysis of the elbow
print("\n" + "=" * 70)
print("ANALYSIS:")
print("=" * 70)
print("""
The "elbow" is where the inertia decrease starts to slow down.
This suggests a good trade-off between model complexity and fit.

In the plot above:
- Sharp decrease from k=2 to k=3 or k=4
- Gradual decrease after that
→ The elbow appears around k=3 or k=4

Our initial choice of k=3 is reasonable!
""")

# Calculate the rate of decrease (difference between consecutive k values)
differences = [inertias[i] - inertias[i+1] for i in range(len(inertias)-1)]
print("\nInertia decrease per additional cluster:")
for i, (k, diff) in enumerate(zip(list(k_values)[:-1], differences)):
    print(f"  k={k} → k={k+1}: decrease of {diff:6.2f}")

print("\n✓ Exercise 2A.2 complete!")
````

#### Exercise 2A.3 ** — Silhouette Score

The silhouette score measures how similar each point is to its own cluster vs. the nearest neighboring cluster. It ranges from -1 (wrong cluster) to +1 (well-clustered).

1. Calculate the silhouette score for your chosen k: `silhouette_score(X_scaled, labels)`
2. Also calculate it for other values of k — does the best silhouette score agree with the elbow?
3. *Optional:* Create a silhouette plot showing per-sample scores

````python
# Exercise 2A.3: Silhouette Score

print("=" * 70)
print("SILHOUETTE SCORE ANALYSIS")
print("=" * 70)

# 1. Calculate silhouette score for different values of k
silhouette_scores = []
k_values = range(2, 11)

print("\nCalculating silhouette scores for k = 2 to 10...")
for k in k_values:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans_temp.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"  k={k:2d}: silhouette score = {score:.4f}")

# 2. Find the best k
best_k_idx = np.argmax(silhouette_scores)
best_k = list(k_values)[best_k_idx]
best_score = silhouette_scores[best_k_idx]

print(f"\n→ Best silhouette score: k={best_k} with score={best_score:.4f}")

# 3. Plot k vs silhouette score
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(k_values, silhouette_scores, 'go-', linewidth=2, markersize=8, label='Silhouette Score')
ax.plot(best_k, best_score, 'r*', markersize=25, label=f'Best: k={best_k}', zorder=5)
ax.plot(3, silhouette_scores[1], 'b^', markersize=12, label='Our choice: k=3', zorder=4)

ax.set_xlabel('Number of Clusters (k)', fontsize=12)
ax.set_ylabel('Silhouette Score', fontsize=12)
ax.set_title('Silhouette Score vs Number of Clusters', fontsize=14, fontweight='bold')
ax.set_ylim([min(silhouette_scores)-0.05, max(silhouette_scores)+0.05])
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11, loc='best')

plt.tight_layout()
plt.show()

# 4. Optional: Create a silhouette plot for our chosen k=3
print("\n" + "=" * 70)
print("SILHOUETTE PLOT FOR k=3")
print("=" * 70)

kmeans_final = KMeans(n_clusters=3, random_state=42, n_init=10)
labels_final = kmeans_final.fit_predict(X_scaled)
silhouette_vals = silhouette_samples(X_scaled, labels_final)

# Create silhouette plot
fig, ax = plt.subplots(figsize=(10, 8))

y_lower = 10
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # Blue, Orange, Green

for i in range(3):
    # Get silhouette values for cluster i
    cluster_silhouette_vals = silhouette_vals[labels_final == i]
    cluster_silhouette_vals.sort()
    
    # Calculate the y-position
    size_cluster_i = cluster_silhouette_vals.shape[0]
    y_upper = y_lower + size_cluster_i
    
    # Plot each cluster's silhouette values
    ax.fill_betweenx(np.arange(y_lower, y_upper),
                      0, cluster_silhouette_vals,
                      facecolor=colors[i], edgecolor=colors[i],
                      alpha=0.7, label=f'Cluster {i}')
    
    y_lower = y_upper + 10

# Add average silhouette line
avg_silhouette = silhouette_vals.mean()
ax.axvline(x=avg_silhouette, color="red", linestyle="--", linewidth=2, 
           label=f'Average: {avg_silhouette:.3f}')

ax.set_xlabel("Silhouette Coefficient", fontsize=12)
ax.set_ylabel("Cluster label", fontsize=12)
ax.set_title(f'Silhouette Plot for K-Means (k=3)', fontsize=14, fontweight='bold')
ax.set_yticks([])
ax.legend(loc='best', fontsize=10)
ax.set_xlim([-0.3, 1])

plt.tight_layout()
plt.show()

# 5. Analysis
print("\n" + "=" * 70)
print("ANALYSIS:")
print("=" * 70)
print(f"""
Silhouette Score Interpretation:
- Range: -1 to +1
- +1: Perfectly clustered (far from neighboring clusters)
-  0: On the border between clusters
- -1: Misclassified (closer to other clusters than its own)

Our Results:
- k=3 silhouette score: {silhouette_scores[1]:.4f}
- Best k according to silhouette: k={best_k} with score {best_score:.4f}
- Average silhouette for k=3: {avg_silhouette:.4f}

Comparison with Elbow Method:
- Elbow method suggested k=3 or k=4
- Silhouette score {'agrees' if best_k in [3, 4] else 'suggests'} with k={best_k}
- Both methods together give us confidence in choosing k=3

Per-cluster quality (silhouette values for k=3):
""")

for i in range(3):
    cluster_vals = silhouette_vals[labels_final == i]
    print(f"  Cluster {i}: min={cluster_vals.min():.3f}, mean={cluster_vals.mean():.3f}, max={cluster_vals.max():.3f}")

print(f"""
→ Clusters with higher values are more cohesive
→ Negative values indicate points that might be misclassified
""")

print("\n✓ Exercise 2A.3 complete!")
````

#### Exercise 2A.4 ** — Interpret the clusters

Clusters are only useful if they *mean something*:

*Important: interpret using the original scale, not the scaled values — those are harder to reason about.*

````python
# Exercise 2A.4: Interpret the clusters

print("=" * 70)
print("CLUSTER INTERPRETATION (using original unscaled data)")
print("=" * 70)

# 1. Group original data by cluster and compute means
df_with_clusters = df.copy()
df_with_clusters['cluster'] = labels_final

print("\nCluster Characteristics (Mean values for each feature):\n")

# Calculate cluster centers using original scale
cluster_means = df_with_clusters.groupby('cluster')[feature_cols].mean()
print(cluster_means)

# 2. Give each cluster a descriptive name
print("\n" + "=" * 70)
print("CLUSTER NAMING")
print("=" * 70)

cluster_names = {}

for cluster_id in range(3):
    cluster_data = cluster_means.loc[cluster_id]
    
    print(f"\nCluster {cluster_id}:")
    print(f"  Bill Length: {cluster_data['bill_length_mm']:.1f} mm")
    print(f"  Bill Depth:  {cluster_data['bill_depth_mm']:.1f} mm")
    print(f"  Flipper Length: {cluster_data['flipper_length_mm']:.1f} mm")
    print(f"  Body Mass: {cluster_data['body_mass_g']:.0f} g")
    
    # Assign descriptive names based on characteristics
    # Using flipper length and body mass as key differentiators
    flipper = cluster_data['flipper_length_mm']
    mass = cluster_data['body_mass_g']
    bill_length = cluster_data['bill_length_mm']
    
    # Create description based on relative sizes
    if mass > 4200:
        size_desc = "Large"
    elif mass > 3700:
        size_desc = "Medium"
    else:
        size_desc = "Small"
    
    if bill_length > 44:
        bill_desc = "Long-billed"
    elif bill_length > 41:
        bill_desc = "Medium-billed"
    else:
        bill_desc = "Short-billed"
    
    cluster_names[cluster_id] = f"{size_desc} {bill_desc} Penguins"
    print(f"  → Description: {cluster_names[cluster_id]}")

# 3. Visualize cluster centers as heatmap
print("\n" + "=" * 70)
print("HEATMAP VISUALIZATION")
print("=" * 70)

# Normalize the cluster means for better heatmap visualization
cluster_means_normalized = (cluster_means - cluster_means.min()) / (cluster_means.max() - cluster_means.min())

fig, ax = plt.subplots(figsize=(10, 5))

# Create heatmap
sns.heatmap(cluster_means_normalized.T, 
            annot=cluster_means.T,  # Show original values
            fmt='.1f',
            cmap='RdYlGn',
            cbar_kws={'label': 'Normalized Value'},
            linewidths=1,
            linecolor='gray',
            ax=ax)

# Customize labels
ax.set_xticklabels([cluster_names[i] for i in range(3)], rotation=45, ha='right')
ax.set_ylabel('Features', fontsize=12)
ax.set_title('Cluster Centers Heatmap (Values shown in original scale)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

# 4. Visualize as bar chart (side by side comparison)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, feature in enumerate(feature_cols):
    ax = axes[idx]
    
    cluster_means[feature].plot(kind='bar', ax=ax, color=['#1f77b4', '#ff7f0e', '#2ca02c'])
    
    ax.set_title(f'{feature}', fontsize=12, fontweight='bold')
    ax.set_xlabel('Cluster', fontsize=11)
    ax.set_ylabel('Mean Value', fontsize=11)
    ax.set_xticklabels([cluster_names[i].split()[0] for i in range(3)], rotation=45, ha='right')
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# 5. Summary comparison table
print("\n" + "=" * 70)
print("SUMMARY TABLE")
print("=" * 70)

summary_df = pd.DataFrame({
    'Cluster': [cluster_names[i] for i in range(3)],
    'Count': [len(df_with_clusters[df_with_clusters['cluster'] == i]) for i in range(3)],
    'Avg Bill Length': cluster_means['bill_length_mm'].round(1),
    'Avg Bill Depth': cluster_means['bill_depth_mm'].round(1),
    'Avg Flipper Length': cluster_means['flipper_length_mm'].round(1),
    'Avg Body Mass': cluster_means['body_mass_g'].round(0)
})

print(summary_df.to_string(index=False))

# 6. Validate against actual species
print("\n" + "=" * 70)
print("VALIDATION: Do clusters match penguin species?")
print("=" * 70)

# Create a crosstab to see if clusters align with species
validation_crosstab = pd.crosstab(df_with_clusters['species'], df_with_clusters['cluster'])
print("\nClusters vs Species (should show clear patterns if good clustering):")
print(validation_crosstab)

print("\n✓ Exercise 2A.4 complete!")
````
